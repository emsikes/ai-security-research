```python
import os
import pprint

from langchain.chat_models import init_chat_model
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage
```

```python
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
model = init_chat_model("gpt-5-mini")
```

```python
# Input - HumanMessage, output AIMessage
model.invoke("Please tell me what is artificial intelligence?")
```

```python
messages = [
    SystemMessage("You are an expert carrer advisor specializing in AI, Agentic AI, Machine Learning, Data Engineering, and Forward Deployed Engineer fields."),
    HumanMessage("Summarize the top 10 most in-demand technical skills for AI and data related skills.")
]

response = model.invoke(messages)
print(response.content)
```

```python
system_message = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_message,
    HumanMessage("How do I create a REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
system_message = SystemMessage("""You are a senior Python developer with expertiese in web frameworks.
You always provide code examples and explain your reasoning.  Be concise but thorough in you explanations.""")

messages = [
    system_message,
    HumanMessage("How do I create a REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
human_msg = HumanMessage(
    content="Hello",
    name="Alice", # Optional - identify user
    id="msg-123", # Optional - unique identifier for tracing
)
```

```python
model.invoke([human_msg])
```

```python
ai_msg = AIMessage("I'd be happy to help you with that question!")

messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great!  What's 2+2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "Los Angeles"},
        "id": "call_123"
    }]
)

weather_result = "Sunny, 87 degrees F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123" # Must match the call ID
)

messages = [
    HumanMessage("What's the weather in Los Angeles?"),
    ai_message, # model's tool call
    tool_message, # tool execution result
]

response = model.invoke(messages)
```

```python
print(response.content)
```
